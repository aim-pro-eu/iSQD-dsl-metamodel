```python
import pandas as pd

competencies = pd.read_excel('EITCI_AIMPRO_USAL_to_EITCI_Mapping_Sheet_v4.xlsx', sheet_name='USAL_to_EITCI_v4')
competencies

def to_jjscript(x):
    return f'''do
create instance of Competency "{x['USAL_ID']}"
set {x['USAL_ID']}.ID = "{x['USAL_ID']}"
set {x['USAL_ID']}.title = "{x['USAL_Competency']}" 
set {x['USAL_ID']}.description = "{x['Description']}" 
set {x['USAL_ID']}.category = "{x['USAL_Category']}" 
set {x['USAL_ID']}.ksa = "{x['Primary_KSA']}" 
set {x['USAL_ID']}.proficiency = "{x['Proficiency_Level']}" 
set {x['USAL_ID']}.heRelevance = "{x['HE_Relevance']}" 
set {x['USAL_ID']}.smeRelevance = "{x['SME_Relevance']}" 
set {x['USAL_ID']}.vetRelevance = "{x['VET_Relevance']}"
set {x['USAL_ID']}.sourceFrameworks = "{x['Source_Frameworks']}"
end
'''

with open('competencies.jjs', 'w', encoding='utf-8') as f:
    f.write('\n'.join(competencies.apply(to_jjscript, axis=1)))
```
