```python
import pandas as pd

competencies = pd.read_excel('EITCI_AIMPRO_USAL_to_EITCI_Mapping_Sheet_v4.xlsx', sheet_name='USAL_to_EITCI_v4')
competencies

TOPIC_TEMPLATE = '''
create instance of Topic "{USAL_ID}"
set {USAL_ID}.ID = "{USAL_ID}"
set {USAL_ID}.title = "{USAL_Competency}"
set {USAL_ID}.description = "{Description}"
set {USAL_ID}.category = "{USAL_Category}"
set {USAL_ID}.ksa = "{Primary_KSA}"
set {USAL_ID}.proficiency = "{Proficiency_Level}"
set {USAL_ID}.heRelevance = "{HE_Relevance}"
set {USAL_ID}.smeRelevance = "{SME_Relevance}"
set {USAL_ID}.vetRelevance = "{VET_Relevance}"
set {USAL_ID}.sourceFrameworks = {Source_Frameworks}
'''

def to_jjscript(x):
    return TOPIC_TEMPLATE.format(
        USAL_ID=x['USAL_ID'],
        USAL_Competency=x['USAL_Competency'],
        Description=x['Description'],
        USAL_Category=x['USAL_Category'].capitalize(),
        Primary_KSA=x['Primary_KSA'].capitalize(),
        Proficiency_Level=x['Proficiency_Level'].capitalize(),
        HE_Relevance=x['HE_Relevance'].capitalize(),
        SME_Relevance=x['SME_Relevance'].capitalize(),
        VET_Relevance=x['VET_Relevance'].capitalize(),
        Source_Frameworks=x['Source_Frameworks'].split(', ')
    )

with open('competencies.jjs', 'w', encoding='utf-8') as f:
    f.write('do\n' + ''.join(competencies.apply(to_jjscript, axis=1)) + '\nend')
```

```python
import yaml

def to_dict(x):
    return {
        'ID': x['USAL_ID'],
        'title': x['USAL_Competency'],
        'description': x['Description'],
        'category': x['USAL_Category'],
        'ksa': x['Primary_KSA'],
        'proficiency': x['Proficiency_Level'],
        'heRelevance': x['HE_Relevance'],
        'smeRelevance': x['SME_Relevance'],
        'vetRelevance': x['VET_Relevance'],
    }

competencies_dict = competencies.apply(to_dict, axis=1).to_list()
yaml_str = yaml.dump(competencies_dict, allow_unicode=True)
print(yaml_str)
```
